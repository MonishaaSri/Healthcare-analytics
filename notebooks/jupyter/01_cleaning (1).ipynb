```python
df.columns = df.columns.str.lower().str.replace(' ', '_')

print("New column names:")
print(list(df.columns))
```

```python
df['date_of_service'] = pd.to_datetime(df['date_of_service'])

df['year']       = df['date_of_service'].dt.year
df['month']      = df['date_of_service'].dt.month
df['month_name'] = df['date_of_service'].dt.strftime('%B')

print("Date range:")
print("From:", df['date_of_service'].min())
print("To:  ", df['date_of_service'].max())
print()
print(df[['date_of_service','year','month','month_name']].head())
```

```python
df['diag_prefix'] = df['diagnosis_code'].str[:3]

icd10_map = {
    'A00': 'Cholera',
    'A01': 'Typhoid and Paratyphoid Fevers',
    'A02': 'Salmonella Infections',
    'A03': 'Shigellosis',
    'A04': 'Bacterial Intestinal Infections',
    'A05': 'Bacterial Foodborne Intoxications',
    'A06': 'Amoebiasis',
    'A07': 'Protozoal Intestinal Diseases',
    'A08': 'Viral Intestinal Infections',
    'A09': 'Gastroenteritis and Colitis',
    'A15': 'Respiratory Tuberculosis',
    'A16': 'Respiratory Tuberculosis (Unconfirmed)',
    'A17': 'Tuberculosis of Nervous System',
    'A18': 'Tuberculosis of Other Organs',
    'A19': 'Miliary Tuberculosis',
}

df['disease_name'] = df['diag_prefix'].map(icd10_map)

print("Disease mapping result:")
print(df['disease_name'].value_counts())
print()
print("Unmapped rows:", df['disease_name'].isnull().sum())
```

```python
df['revenue_leakage'] = df['billed_amount'] - df['paid_amount']

df['reimbursement_rate'] = round((df['paid_amount'] / df['billed_amount']) * 100, 2)

df['is_denied'] = np.where(df['outcome'] == 'Denied', 1, 0)

print("New columns added successfully!")
print()
print(df[['billed_amount','paid_amount','revenue_leakage','reimbursement_rate','is_denied']].head(10))
print()
print("Average reimbursement rate:",  round(df['reimbursement_rate'].mean(), 2), "%")
print("Total revenue leakage: $", df['revenue_leakage'].sum())
```

```python
# FIX 1 — Convert date_of_service to proper date
df['date_of_service'] = pd.to_datetime(df['date_of_service']).dt.date

# FIX 2 — Rename follow-up_required (hyphen to underscore)
df = df.rename(columns={'follow-up_required': 'follow_up_required'})

# Verify both fixes
print("Date type now:", df['date_of_service'].dtype)
print()
print("Column names:")
for col in df.columns:
    print(" -", col)
```

```python
# First fix column names to make sure they are all lowercase
df.columns = df.columns.str.lower().str.replace(' ', '_')

# Verify columns are correct
print("Current columns:", list(df.columns))
print()

# -----------------------------------------------
# CREATE 4 CHARTS IN ONE IMAGE
# -----------------------------------------------
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Healthcare Claims Analysis', fontsize=16, fontweight='bold')

# -----------------------------------------------
# CHART 1 — Top Left — Claims by Outcome
# -----------------------------------------------
outcome_counts = df['outcome'].value_counts()
axes[0,0].bar(
    outcome_counts.index,
    outcome_counts.values,
    color=['green', 'red', 'orange']
)
axes[0,0].set_title('Claims by Outcome')
axes[0,0].set_xlabel('Outcome')
axes[0,0].set_ylabel('Number of Claims')

# -----------------------------------------------
# CHART 2 — Top Right — Insurance Type
# -----------------------------------------------
insurance_counts = df['insurance_type'].value_counts()
axes[0,1].pie(
    insurance_counts.values,
    labels=insurance_counts.index,
    autopct='%1.1f%%'
)
axes[0,1].set_title('Claims by Insurance Type')

# -----------------------------------------------
# CHART 3 — Bottom Left — Top Diseases
# -----------------------------------------------
disease_counts = df['disease_name'].value_counts().head(10)
axes[1,0].barh(
    disease_counts.index,
    disease_counts.values,
    color='steelblue'
)
axes[1,0].set_title('Top 10 Diseases by Claim Count')
axes[1,0].set_xlabel('Number of Claims')

# -----------------------------------------------
# CHART 4 — Bottom Right — Monthly Trend
# -----------------------------------------------
monthly = df.groupby('month')['claim_id'].count()
axes[1,1].plot(
    monthly.index,
    monthly.values,
    marker='o',
    color='purple',
    linewidth=2
)
axes[1,1].set_title('Monthly Claims Trend')
axes[1,1].set_xlabel('Month')
axes[1,1].set_ylabel('Number of Claims')
axes[1,1].set_xticks([5, 6, 7, 8, 9])
axes[1,1].set_xticklabels(['May', 'Jun', 'Jul', 'Aug', 'Sep'])

# -----------------------------------------------
# SAVE AND SHOW
# -----------------------------------------------
plt.tight_layout()

# ✅ UPDATED SAVE PATH
save_folder = r"C:\Users\Monishaa Sri .P\OneDrive\Documents\healthcare analytics\notebooks\cleaned"
os.makedirs(save_folder, exist_ok=True)

plt.savefig(save_folder + r"\medical_charts.png", dpi=150)
plt.show()

print("✅ Charts saved successfully!")
print("Location:", save_folder)
```
